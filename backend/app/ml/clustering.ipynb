```python
from langchain_community.document_loaders import CSVLoader
import os
```

```python

path_csv = "/Users/lait-zet/Desktop/ragprojects/Document/it_support_questions.csv"
print(path_csv)
laod = CSVLoader(path_csv)
document = laod.load()
texts = [doc.page_content for doc in document]
for i in range(0 ,2):
    print(texts[i])
```

```python
from langchain_huggingface import HuggingFaceEmbeddings

embedding = HuggingFaceEmbeddings(model_name="sentence-transformers/all-MiniLM-L6-v2")

X = await embedding.aembed_documents(texts)
print(X[0])
```

Train The model 

```python
from sklearn.cluster import KMeans

k  = 5 
kmeans = KMeans(n_clusters=k, random_state=0)
kmeans.fit(X)

labels = kmeans.labels_
```

```python
for i in range(k):
    print(f"Cluster {i}:")
    for j, label in enumerate(labels):
        if label == i:
            pass
            # print(f" - {texts[j]}")
```

```python
#save model
import joblib
model_path = "/Users/lait-zet/Desktop/ragprojects/backend/app/ml/kmeans_model.pkl"
joblib.dump(kmeans, model_path)
print(f"Model saved to {model_path}")
```

```python
query = "what is computer"
embedding = HuggingFaceEmbeddings(model_name="sentence-transformers/all-MiniLM-L6-v2")
loaded_model = joblib.load(model_path)
query_embedding = await embedding.aembed_query(query)
cluster_number = int(loaded_model.predict([query_embedding])[0])
print(type(cluster_number))
```
